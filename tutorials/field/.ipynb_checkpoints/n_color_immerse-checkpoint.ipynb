# `n_color_immerse()` — director-to-RGB color mapping

`nematics3d.field.n_color_immerse()` maps a nematic director field to RGB colors.
This notebook explains what the function does, why it is designed the way it is,
and how to call it directly.

## Background: why director coloring needs a special scheme

A nematic director **n** is an unoriented line: **n** and **−n** represent
the same physical state. Any color assigned to a director must therefore
satisfy `color(n) == color(−n)`, which rules out all standard vector colormaps
(HSV, rainbow, etc.) because those treat opposite directions as maximally different.

The natural mathematical space for all unoriented unit vectors in 3D is the
real projective plane **RP²** — the unit sphere S² with antipodal points identified.
What we need is a continuous map from RP² to a valid RGB cube [0, 1]³.

## Design goals

The mapping in `n_color_immerse()` is designed to satisfy four properties,
in rough order of priority:

1. **Continuity** — nearby director orientations map to similar colors.
   This is the minimum requirement for any useful colormap on RP².

2. **Injectivity where possible** — different orientations map to different colors
   as much as the topology allows. (A perfectly injective continuous map from RP² to
   R³ does not exist, so the best one can do is minimize collisions.)

3. **Distance from white** — all colors are kept away from `(1, 1, 1)`,
   so every director is clearly visible on a white background.

4. **Axis anchoring** — the three coordinate axes are mapped close to the
   three primary colors: **X → red**, **Y → blue**, **Z → green**.
   This makes it easy to read off the dominant orientation in a scene.

## Implementation

The mapping is derived from **Boy's surface**, a specific smooth immersion of RP²
into R³. The Boy's surface parameterization provides a polynomial map
from (x, y, z) on the unit sphere to a point in R³ that is the same for
antipodal pairs by construction. `n_color_immerse()` takes this polynomial,
applies a fixed linear transformation and per-channel scaling, and clips the
result to [0, 1]³ to produce valid RGB values.

The exact colors for the six axis directions are:

| Direction | Approximate RGB |
| --- | --- |
| ±X | `(0.91, 0.23, 0.22)` — red |
| ±Y | `(0.05, 0.28, 0.49)` — blue |
| ±Z | `(0.30, 0.90, 0.23)` — green |

## Usage

The function accepts a director field of shape `(..., 3)` — any number of leading
dimensions — and returns a list of RGB tuples with the same leading shape.

````python
import sys
from pathlib import Path

import numpy as np

REPO_ROOT = next(
    p for p in [Path.cwd(), Path.cwd().parent]
    if (p / 'src' / 'nematics3d').exists()
)
try:
    from nematics3d.field import n_color_immerse
except ModuleNotFoundError:
    sys.path.insert(0, str(REPO_ROOT / 'src'))
    from nematics3d.field import n_color_immerse

# Color a few axis-aligned directors
n = np.array([
    [1, 0, 0],   # +X
    [-1, 0, 0],  # -X  (should match +X)
    [0, 1, 0],   # +Y
    [0, 0, 1],   # +Z
])

colors = n_color_immerse(n)
for director, color in zip(n, colors):
    print(f'n={director}  ->  RGB={tuple(round(c, 4) for c in color)}')
````

The first two rows (`+X` and `−X`) print identical RGB values,
confirming that the antipodal symmetry `color(n) == color(−n)` holds.

## Visualizing the full colormap: the color sphere

The most convenient way to see the entire colormap at once is the **color sphere**:
a unit sphere whose surface is colored by applying `n_color_immerse()` to each
vertex normal. Because antipodal points share a color, opposite hemispheres look
identical.

If you already have a `PlotFigure` object `figure`, call:

```python
figure.act_add_axes_widget(style='sphere')
```

This adds the color sphere as a corner orientation widget. As you rotate the
main view, the sphere rotates with it, so its visible face always shows the
color corresponding to the current viewing direction.
