# `fmt_value()`

`nematics3d.format.fmt_value()` formats a real scalar or a real `NumPy` array as compact fixed-decimal text for messages, labels, and other human-readable output.

## What `fmt_value()` is for

Use `fmt_value()` when numerical values need a predictable, compact textual representation rather than the default Python or `NumPy` representation. It is especially useful when a scalar, vector, or small array is embedded inside a longer message.

The function applies the same fixed number of decimal places to every displayed value. One-dimensional arrays are kept on a single line so vectors remain convenient to embed in interactive messages; higher-dimensional arrays preserve their array structure.

## Setup

**For readers who are only interested in the tutorial, this section can be safely skipped.** The following cell only imports `NumPy` and `fmt_value()`.

````python
import numpy as np

from nematics3d.format import fmt_value
````

## Minimal example

Pass a real scalar or real `NumPy` array as `v`. By default, each value is displayed with two digits after the decimal point.

````python
value = np.array([1.2345, 2.0, -0.4567])
text = fmt_value(value)

print(text)
````

### What is returned

`fmt_value()` always returns a Python `str`. For the example above, the result is:

```text
[1.23, 2.00, -0.46]
```

The input array is not modified.

## Arguments

The public signature is:

```python
fmt_value(v, ndigits=2)
```

| Argument | What it controls | Typical form |
| --- | --- | --- |
| `v` | The value to format. It must be a real scalar or a real `NumPy` array. Boolean, complex, string, and other non-real inputs are rejected. | `1.25`, `np.float64(1.25)`, or `np.array([...])` |
| `ndigits=2` | Number of digits displayed after the decimal point. It must be a non-negative Python or `NumPy` integer; booleans are rejected. | `0`, `2`, `4` |

### Example: change the number of decimal places

Use `ndigits` when the surrounding message needs more or less precision. The formatting is fixed-point and zero-padded.

````python
value = 1.23456

print(fmt_value(value, ndigits=0))
print(fmt_value(value, ndigits=2))
print(fmt_value(value, ndigits=4))
````

## Special examples


### Special case: one-dimensional arrays stay on one line

One-dimensional arrays are deliberately formatted as a single line. This behavior is useful for vectors embedded in GUI messages or other short text, and does not depend on `NumPy`'s normal line-wrapping width.

````python
vector = np.linspace(0.0, 1.0, 12)
text = fmt_value(vector)

print(text)
print("contains newline:", "\n" in text)
````

### Special case: multidimensional arrays preserve structure

For arrays with two or more dimensions, `fmt_value()` uses `NumPy`'s structured array display instead of flattening the data. This keeps rows and higher-dimensional blocks recognizable.

````python
matrix = np.array([[1.0, 2.345], [3.456, -4.0]])
print(fmt_value(matrix))
````

### Special case: `NaN` and infinity

`NaN` and positive or negative infinity are real floating-point values and are preserved as `nan`, `inf`, and `-inf` in the formatted text.

````python
special = np.array([np.nan, np.inf, -np.inf])
print(fmt_value(special))
````

## Usage tips

### Embed the result directly in a message

Because the return value is already a string, it can be inserted directly into an f-string. This is the main pattern used by `Nematics3D` visualization code when reporting a picked position, orientation, tangent, color, or scalar value.

````python
orientation = np.array([0.1234, 0.9876, -0.4321])
message = f"Local orientation: {fmt_value(orientation)}"
print(message)
````

## Details

`fmt_value()` is a display-formatting helper rather than a numerical transformation. Scalar values are converted to a Python `float` and formatted with fixed-point notation. A zero-dimensional array follows the same scalar path.

For a one-dimensional array, each element is formatted individually and joined explicitly so the complete vector stays on one line. For arrays with two or more dimensions, the numeric array is passed to `numpy.array2string()` with a fixed-point formatter. This preserves the array structure and allows `NumPy` to summarize large multidimensional arrays without first constructing a full array of formatted strings.

The function is intentionally strict about its input domain. It is intended for real numerical display, so booleans, complex values, strings, and non-real arrays raise `TypeError` instead of silently falling back to an unrelated string representation. Invalid values of `ndigits` also raise an error rather than changing the formatting behavior implicitly.

## Summary

- Use `fmt_value()` to produce compact fixed-decimal text from real scalars and real `NumPy` arrays.
- The default is two digits after the decimal point; change this with `ndigits`.
- One-dimensional arrays deliberately stay on one line, while multidimensional arrays preserve their structure.
- The function returns a string and does not modify the input.
- Non-real inputs and invalid precision arguments are rejected explicitly.
