# 02 — Interacting with figures

In Tutorial 01, we used `quick_visualize_q()` to turn a $Q$-tensor field into an interactive 3D view. Now we will learn how to explore and modify that figure interactively.

`Nematics3D` figures are designed for exploration rather than passive viewing. You can rotate the scene, select visual objects directly with the mouse, and change their most useful display options without rebuilding the plot.


## Recreate the figure from Tutorial 01

We begin with exactly the same example $Q$ field and the same one-line visualization workflow.


```python
import numpy as np
import nematics3d as n3d

Q = np.load("../../example/data/Q_example_workflow.npy")
Q_obj, figure = n3d.quick_visualize_q(Q=Q)
```

## Start with the mouse

The basic 3D controls are the familiar `PyVista` controls:

- **Left drag** — rotate the scene.
- **Middle drag** — pan the scene.
- **Right drag** — zoom continuously.
- **Mouse wheel** — zoom in or out.

Try these first. They are enough to inspect the overall geometry from different viewpoints.


## Select something in the figure

The more useful interaction begins when you select a plotted object.

**Right-click** a glyph in the scene. `Nematics3D` will highlight or dehighlight that object and print its object information in the figure console.

For example, try selecting the director rods on the plane:

![Selecting a glyph](../output/interact_with_figures/right_single.png)

The important idea is that the objects in a `Nematics3D` figure are not merely pixels. The figure still knows which plotted object you clicked, so it can identify and interact with that object directly.


## Change the selected object

To open the interactive control panel for a glyph, **double-right-click** it. If that object supports interactive controls, `Nematics3D` will open its panel.

For a rod glyph, for example, the panel exposes frequently used display options such as its radius and length.

![Glyph options](../output/interact_with_figures/figure_opts.png)

Change one of these values and watch the figure update. You do not need to rerun the visualization from scratch: the plotted object updates in place.

This live connection between scientific objects and their visualization is an important part of the `Nematics3D` workflow. We will return to the object model behind it in later tutorials.


## Select more than one object

Sometimes several plotted objects should be adjusted together.

After selecting the first object, hold **Ctrl + Shift** and **right-click** another one. The new object is added to the current selection instead of replacing it.

![Selecting multiple glyphs](../output/interact_with_figures/right_double.png)

The control panel now applies to the selected group where the relevant option is shared. This is useful when several visual elements should use the same scale or appearance.


## Remove an object from the figure

If a visual element is getting in the way, you can remove it directly from the interactive scene.

Hold **Ctrl** and **double-left-click** the glyph.

![Removing a glyph](../output/interact_with_figures/left_double.png)

This is useful during exploration when you want to simplify a crowded scene and focus on the structures that matter.
