# 04 — Working with a `QFieldObject`

Tutorial 01 gave us `Q_obj`, and Tutorial 03 taught us how to inspect an unfamiliar `Nematics3D` object. Now we can ask a more useful question: **what is `Q_obj` actually for?**

A `QFieldObject` is the central analysis object for a $Q$-tensor field. It keeps the field together with quantities derived from it, defects and disclination lines detected from it, and scientific objects created for later analysis and visualization.


## Start from the same $Q$ field


````python
import numpy as np
import nematics3d as n3d

Q = np.load("../../example/data/Q_example_workflow.npy")
Q_obj, figure = n3d.quick_visualize_q(Q=Q)
````

## From $Q$ to $S$ and $\mathbf{n}$

The array we supplied contains the tensor field $Q$. A `QFieldObject` also exposes the two quantities we most often want to interpret directly: the scalar order parameter $S$ and director $\mathbf{n}$.


````python
Q_obj.S.shape, Q_obj.n.shape
````

This is the first useful mental model for `QFieldObject`: it is not merely a wrapper around the input array. It gives one object from which we can reach the field and its standard derived representations.


## The defect lines belong to the same analysis object

The colored curves shown in Tutorial 01 were not produced by an unrelated plotting routine. During the quick workflow, `QFieldObject` detected defects and classified them into disclination lines. Those lines remain available from the same object.


````python
len(Q_obj.lines)
````

`QFieldObject` also keeps physical and geometric objects derived from the field in its object registry. We already encountered one of them in Tutorial 03: the director plane created by `quick_visualize_q()`.


````python
Q_obj.objs
n_plane = Q_obj.objs["n-plane"]
n_plane
````

## Ask the field object to create another analysis view

Because `QFieldObject` is the entry point for operations on this field, its `act_...` methods let us move from stored data to new analysis objects. For example, we can create another director plane on a new figure:


````python
Q_obj.act_visualize_n_plane(
    is_new=True,
    plane_name="second-n-plane",
)
````

The important point is not the particular plane. It is the workflow:

```text
$Q$-tensor data
      ↓
QFieldObject
      ↓
derived fields, defects, lines, and analysis objects
      ↓
interactive visualization
```

This is why `Q_obj` is worth keeping after `quick_visualize_q()` returns. It is the object through which later analysis of the same $Q$ field can continue.


## You do not need to memorize the rest

`QFieldObject` has many more operations: interpolation, defect analysis, line smoothing, scalar-order planes, near-defect director views, and more. A guided introduction does not need to list them all. Tutorial 03 already gave us the discovery tools:

```python
Q_obj.show_     # inspect and explain
Q_obj.act_      # discover available actions
Q_obj.show_relations()
```

Use those tools when the next scientific question requires them.


## The main idea

A `QFieldObject` is the central working object for one $Q$-tensor field. From it you can reach $S$ and $\mathbf{n}$, inspect detected disclination lines, access derived scientific objects, and launch further analysis and visualization.

There is one design feature we have now encountered repeatedly but have not explained. Many `Nematics3D` objects have configurable options, and changing those options can immediately change an existing result or visualization. Tutorial 05 explains the common mechanism behind that behavior.
