# 03 — How to understand an unfamiliar `Nematics3D` object?

In Tutorial 02, right-clicking a visual element revealed something important: a `Nematics3D` figure is not just a collection of pixels. The things you see in the figure are connected to actual Python objects that `Nematics3D` can identify and describe.

We have already had one such object since Tutorial 01. The call to `quick_visualize_q()` returned both the figure and `Q_obj`. Until now, we have mostly ignored `Q_obj`.

So the next question is: **if `Nematics3D` gives us an unfamiliar object, how do we find out what it is and what we can do with it?**


## Recreate the object from Tutorial 01

We begin with the same example $Q$-tensor field and the same visualization call. This time, our attention is on the returned object rather than the figure.


````python
import numpy as np
import nematics3d as n3d

Q = np.load("../../example/data/Q_example_workflow.npy")
Q_obj, figure = n3d.quick_visualize_q(Q=Q)

Q_obj
````

## Start by asking the object

Suppose we do not know the API of `Q_obj`. A useful first move in Jupyter, IPython, or an editor with autocomplete is to type:

```python
Q_obj.show_
```

and press **Tab** or invoke autocomplete.

The `show_` prefix groups methods whose purpose is to inspect, display, or explain the object. Instead of starting with the source code or memorizing a large API, we can let the object tell us about itself.


## What are you?

The first question is the simplest one:


````python
Q_obj.show_doc()
````

`show_doc()` displays the documentation of the concrete class of the object. Here it explains the role of the object representing our $Q$-tensor field.

This is usually the fastest way to understand an unfamiliar object's overall purpose before looking at individual attributes or methods.


## What information do you contain?

Once we know the object's role, the next question is what information it exposes:


````python
Q_obj.show_readable_attrs()
````

This is deliberately more useful than dumping every Python implementation detail. It presents the attributes that belong to the object's readable interface together with descriptions of what they mean.

If one attribute catches your attention, you can ask about that field specifically. For example, if autocomplete or the list above shows an attribute you want to understand, use:

```python
Q_obj.show_attr_doc("attribute_name")
Q_obj.show_attr_info("attribute_name")
```

`show_attr_doc(...)` answers what the attribute means. `show_attr_info(...)` gives a fuller description, including information such as its category, current value, aliases, and whether it can be modified.


## The names themselves carry information

The usefulness of typing `show_` is not accidental. Many `Nematics3D` objects follow a common naming vocabulary so that autocomplete can be used as an exploration tool.

For methods, two prefixes are especially useful:

| Prefix | Meaning | Question to ask |
| --- | --- | --- |
| `show_...` | inspect, display, or explain | What can you tell me? |
| `act_...` | perform an action | What can you do? |

So if you want to discover available operations, you do not need to know their full names in advance:

```python
Q_obj.show_    # autocomplete: inspection and explanation
Q_obj.act_     # autocomplete: actions
```


### Attribute prefixes

Structured attributes use the same idea. Their prefixes indicate the role they play inside the object:

| Prefix | Meaning | A useful way to read it |
| --- | --- | --- |
| `raw_...` | canonical stored input or base data | What was given to the object? |
| `state_...` | writable runtime state | What state is it currently in? |
| `default_...` | managed default-layer input | What default value does it carry? |
| `calc_...` | computed readable data | What has it calculated? |
| `entity_...` | generated object-valued result | What object has it created? |

This makes autocomplete useful even when you remember only the *kind* of information you want:

```python
Q_obj.raw_
Q_obj.state_
Q_obj.default_
Q_obj.calc_
Q_obj.entity_
```

You are narrowing the interface by meaning rather than searching through an undifferentiated list of names.


## Objects are connected to other objects

Tutorial 02 already hinted that the visualization is connected to the scientific objects behind it. These connections can also be inspected explicitly.

The figure created an `n-plane` object for visualizing the director field. We can retrieve it from the $Q$-field object and ask what it is related to:


````python
n_plane = Q_obj.objs["n-plane"]
n_plane.show_relations()
````

`show_relations()` shows the objects directly connected to `n_plane`: for example its grid, interpolator, registry, and visual objects.

If you want to follow those connections beyond the first layer, use:

```python
n_plane.show_relation_tree(depth=3)
```

This gives us two complementary ways to understand an unfamiliar object: **attributes tell us what the object contains; relations tell us where it sits in the larger object system.**


## Why do different objects behave this way?

We have now used the same discovery vocabulary on scientific and visualization-related objects without first learning each class separately. That consistency comes from a common base object protocol.

At the bottom of that protocol is `ClassBase`. Users normally do not instantiate `ClassBase` directly. Instead, many concrete `Nematics3D` classes inherit its conventions for self-description, attribute organization, actions, and object relations.

For ordinary use, the important thing is therefore not the inheritance hierarchy itself. It is the vocabulary that the hierarchy gives you.


## A practical discovery workflow

When `Nematics3D` gives you an object you have never used before, this is a useful starting sequence:

```python
obj.show_                     # autocomplete: how can it explain itself?
obj.show_doc()                # what is it?
obj.show_readable_attrs()     # what does it contain?
obj.show_attr_info("...")   # what exactly is this field?

obj.raw_                      # base inputs
obj.state_                    # runtime state
obj.default_                  # default-layer values
obj.calc_                     # calculated data
obj.entity_                   # generated objects
obj.act_                      # available actions
```

For an object that participates in a larger structure, also try `show_relations()` and `show_relation_tree()`.

The main habit is simple: **do not begin by memorizing every class. Learn the common vocabulary, then let autocomplete and the object itself tell you what this particular object provides.**
