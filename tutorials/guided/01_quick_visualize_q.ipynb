# 01 — Your first look at a $Q$-tensor field

`Nematics3D` is a Python package for working with three-dimensional liquid-crystal data. It provides tools for analyzing fields and geometric structures, visualizing them in 3D, and continuing to interact with the resulting scientific objects.

But when you first receive a new dataset, detailed analysis is usually not the first thing you want to do. A much more immediate question is:

> **What does this $Q$-tensor field look like?**

You may want to see the overall director structure, locate the major disclination lines, check whether the data look reasonable, or simply decide what deserves a closer look.

For this reason, one of the most useful entry points to `Nematics3D` is `quick_visualize_q()`. It turns a $Q$-tensor field into a useful interactive 3D view with a single function call. This is often the first function a new user needs, and it remains useful as a quick inspection tool in everyday work.


## Load an example $Q$ field

We will use the example $Q$-tensor field included with the repository.


```python
import numpy as np
import nematics3d as n3d

Q = np.load("../../example/data/Q_example_workflow.npy")
Q.shape
```

## Take a first look

Once the $Q$ field is available, the shortest useful `Nematics3D` workflow is simply:


```python
Q_obj, figure = n3d.quick_visualize_q(Q=Q)
```

That is enough to produce an interactive overview of the dataset. By default, `Nematics3D` constructs a $Q$-field object, detects and visualizes the disclination lines, draws the simulation boundary, and places a director plane through the sample.

A typical result looks like this:

![Default quick visualization](../output/quick_visualize_q/quick_py_main_preview.png)

The rods show the local nematic director on a plane through the volume. The colored curves show the detected disclination lines. For many quick checks, this one call is all you need.
