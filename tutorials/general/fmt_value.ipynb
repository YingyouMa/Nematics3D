# `fmt_value()`

`fmt_value()` formats real scalars and real NumPy arrays as fixed-decimal text.

## Setup

**For readers who are only interested in the tutorial, this section can be safely skipped.**

````python
import numpy as np

from nematics3d.general.format import fmt_value
````

## Minimal example

By default, arrays follow NumPy's normal display behavior, including line wrapping and summarization for sufficiently large arrays.

````python
value = np.array([1.2345, 2.0, -0.4567])
print(fmt_value(value))
````

## Arguments

```python
fmt_value(v, ndigits=2, is_1d_single_line=False)
```

- `v`: real scalar or real NumPy array.
- `ndigits`: non-negative integer controlling digits after the decimal point.
- `is_1d_single_line`: when `True`, a one-dimensional array is fully rendered on one line. The default is `False`, so one-dimensional arrays behave like other NumPy arrays.

### Change the number of decimal places

````python
value = 1.23456
print(fmt_value(value, ndigits=0))
print(fmt_value(value, ndigits=2))
print(fmt_value(value, ndigits=4))
````

## Special examples

### Keep a one-dimensional array on one line

Set `is_1d_single_line=True` when a one-dimensional vector must remain complete and on one line, for example when embedding it in an interactive message. This opt-in mode does not summarize the array with `...`.

````python
vector = np.arange(100, dtype=float) / 10
text = fmt_value(vector, ndigits=1, is_1d_single_line=True)

print(text)
print("contains newline:", "\n" in text)
print("contains omission marker:", "..." in text)
````

### Large multidimensional arrays are summarized

Multidimensional arrays use `numpy.array2string()`. Large arrays therefore follow NumPy's normal summarization behavior and omit undisplayed values with `...` instead of constructing a string containing every element.

````python
large_array = np.arange(10_000, dtype=float).reshape(100, 100)
print(fmt_value(large_array, ndigits=1))
````

### `NaN` and infinity

````python
special = np.array([np.nan, np.inf, -np.inf])
print(fmt_value(special))
````

## Details

Zero-dimensional arrays follow the scalar path. By default, arrays with one or more dimensions are formatted through `numpy.array2string()`, preserving NumPy's structured layout and summarization behavior. The only exception is a one-dimensional array with `is_1d_single_line=True`: every element is then formatted explicitly and joined into one line.

The function accepts real numerical values only. Boolean, complex, string, and other non-real inputs are rejected explicitly.

## Summary

- `fmt_value()` uses fixed-decimal formatting.
- Arrays use NumPy's normal display and summarization behavior by default.
- Set `is_1d_single_line=True` only when a complete one-dimensional array must stay on one line.
- Large multidimensional arrays are summarized with `...`.
- The function returns a string and does not modify the input.
